### Bước 2: Tải file giọng mẫu của bạn lên Colab
- Bấm vào biểu tượng thư mục ở thanh bên trái của Colab.
- Kéo thả file video (`.mp4`) hoặc audio (`.wav`, `.mp3`, `.m4a`) nói giọng của bạn vào đó.
- Đổi tên file thành `my_voice.mp4` hoặc `my_voice.wav`.

```python
# Bước 3: Tự động trích xuất âm thanh từ Video thành file WAV chuẩn
import os

if os.path.exists('my_voice.mp4'):
    !ffmpeg -y -i my_voice.mp4 -vn -acodec pcm_s16le -ar 24000 -ac 1 voice_sample.wav
elif os.path.exists('my_voice.wav'):
    !ffmpeg -y -i my_voice.wav -acodec pcm_s16le -ar 24000 -ac 1 voice_sample.wav
elif os.path.exists('my_voice.mp3'):
    !ffmpeg -y -i my_voice.mp3 -acodec pcm_s16le -ar 24000 -ac 1 voice_sample.wav
else:
    print('⚠️ Chưa tìm thấy file my_voice. Vui lòng tải file lên Colab!')

if os.path.exists('voice_sample.wav'):
    print('✅ File giọng mẫu đã sẵn sàng!')
```

```python
# Bước 4: Nhập kịch bản cần đọc và thực hiện Clone giọng đọc bài!
# Bạn có thể dán đoạn thoại từ Audio 1 -> Audio 5 vào bên dưới:

text_to_read = '''
Hey Brad, ... are you ready to leave? ... Did you peel the apples ... and put the peel in the bin?
Oh no, ... stop it leaking! ... Do not just let the dog lick it! ... Take a pin or a peg to fix it.
Wait, ... do we need bread or jam?
Got it. ... Should I bring a pen to write down the bill, ... or maybe a cheque?
Okay. ... By the way, ... where are we going to live during the trip? ... Near the sheep or near the big ship?
Awesome! ... Ring the bell, ... mark an X on the map, ... and let us go!
'''

from f5_tts.api import F5TTS

f5tts = F5TTS()
output_file = 'Bai_Nop_Giong_Toi_Audio_1.wav'

f5tts.infer(
    ref_file='voice_sample.wav',
    ref_text='',  # Không bắt buộc nhập transcript của mẫu
    gen_text=text_to_read.strip(),
    file_wave=output_file,
    speed=0.90   # Tốc độ đọc 0.90x kiểu học viên
)

# Tinh chỉnh mô phỏng mic tai nghe tự nhiên
!ffmpeg -y -i Bai_Nop_Giong_Toi_Audio_1.wav -filter:a 'highpass=f=100,lowpass=f=7000' Bai_Nop_Hoan_Chinh.mp3
print('🎉 ĐÃ XONG! Tải file Bai_Nop_Hoan_Chinh.mp3 về máy để nộp bài!')
```
